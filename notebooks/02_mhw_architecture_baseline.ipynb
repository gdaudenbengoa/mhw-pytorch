## Module 02: Neural Network Baseline for Marine Heatwave Forecasting

### Objective
In this second module, we will design and build our first neural network architecture in PyTorch to process the spatiotemporal Sea Surface Temperature (SST) tensors we prepared in Module 01. 

Since our input data has a 4D shape of `[Batch_Size, Time_Channels, Height, Width]`, standard fully connected networks (Linear layers) are not suitable because they destroy spatial relationships. Instead, we will build a baseline Convolutional Neural Network (CNN) using 2D convolutions (`nn.Conv2d`) adapted to handle temporal sequences as input channels.

### Architecture Overview
1. **Input Layer**: Receives historical SST maps over $N_{steps} = 5$ days (Channels = 5, Grid = $32 \times 32$).
2. **Feature Extraction**: Uses convolutional layers to learn spatial patterns and local anomalies associated with Marine Heatwaves (MHW).
3. **Output Layer**: Predicts a single-channel anomaly map for the target day $t+5$ (Shape: `[Batch_Size, 1, 32, 32]`).

```python
"""
Module 02 - Cell 1: Baseline Convolutional Neural Network (CNN) for MHW
----------------------------------------------------------------------
Defines a simple PyTorch neural network to process 4D spatiotemporal tensors 
(SST historical sequences) and predict future marine heatwave conditions.
"""
# Import necessary libraries
import torch
import torch.nn as nn

# Allows to register the weights of the model and move it to GPU if available
class MHWBaselineCNN(nn.Module):
    """
    A baseline 2D Convolutional Neural Network for SST forecasting.
    Takes an input tensor of shape [Batch, Time_Channels, Height, Width] 
    and outputs a predicted anomaly map of shape [Batch, 1, Height, Width].
    """
    # The constructor and the forward method define the architecture and data flow through the network.
    def __init__(self, in_channels=5):
        super(MHWBaselineCNN, self).__init__()
        
        # Feature extraction layers using 2D convolutions
        self.conv1 = nn.Conv2d(in_channels=in_channels, out_channels=16, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv2d(in_channels=16, out_channels=1, kernel_size=3, padding=1)

    # The data flow    
    def forward(self, x):
        """
        Forward pass: defines how data flows through the network layers.
        """
        # Pass through first convolution and activation function
        x = self.conv1(x)
        x = self.relu(x)
        
        # Pass through final convolution to get the 1-channel prediction map
        x = self.conv2(x)
        
        return x

# --- Verification & Testing the Architecture ---
if __name__ == "__main__":
    # 1. Instantiate the model
    model = MHWBaselineCNN(in_channels=5)
    
    # 2. Create a dummy input tensor simulating a mini-batch of 4 samples 
    # (Batch=4, Time_Channels=5, Height=32, Width=32)
    dummy_input = torch.randn(4, 5, 32, 32)
    
    # 3. Pass the dummy input through the model
    output = model(dummy_input)
    
    print("--- Cell 1 Output ---")
    print(f"Model Architecture:\n{model}")
    print(f"Input tensor shape:  {dummy_input.shape}")
    print(f"Output tensor shape: {output.shape}")
```
