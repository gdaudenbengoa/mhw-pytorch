# Module 03: Training Loop and Optimization Strategy

## Objective
In this module, we construct the complete training pipeline for our Marine Heatwave baseline model (`MHWBaselineCNN`). We will connect the 4D spatiotemporal tensors generated by our `DataLoader` from Module 01 to the baseline CNN designed in Module 02.

## Key Components
1. **Loss Function**: Mean Squared Error (`nn.MSELoss`) to quantify pixel-wise SST prediction errors.
2. **Optimizer**: Adam (`torch.optim.Adam`) for adaptive learning rate optimization.
3. **Training Loop Pipeline**:
   - Forward pass: Compute predictions $\hat{Y}$ from historical inputs $X$.
   - Loss calculation: Compare $\hat{Y}$ against target SST ground truth $Y$.
   - Backward pass: Calculate gradients via backpropagation (`loss.backward()`).
   - Parameter update: Adjust network weights (`optimizer.step()`).
   - Zero gradients: Clear gradient accumulators (`optimizer.zero_grad()`).
4. **Model Checkpointing**: Save the trained model parameters (`.pth` file).

### Module 03 - Cell 1: Environment Path Setup & Modular Import
In this cell, we configure Python's path to correctly locate our custom `src/` directory and import the `MHWBaselineCNN` architecture defined in Module 2.

```python
import sys
import os

# Ensure the project root path is included so Python can find the 'src' folder
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.append(project_root)

import torch
import torch.nn as nn
import torch.optim as optim

# Import the modular architecture from the src/ folder
from src.models import MHWBaselineCNN

print("--- Path and Module Import Successful ---")
```

### Module 03 - Cell 2: Model Initialization, Loss Function, and Optimizer
Here we instantiate our baseline CNN model, define the Mean Squared Error (MSE) loss function for spatial temperature regression, and set up the Adam optimizer.

```python
# 1. Instantiate the model with the correct historical time channels (5 days)
model = MHWBaselineCNN(in_channels=5)

# 2. Define the Loss Function (MSE for spatial temperature regression)
criterion = nn.MSELoss()

# 3. Define the Optimizer (Adam algorithm for adaptive gradient descent)
learning_rate = 0.001
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

print("--- Model Setup Completed ---")
print(model)
print(f"\nLoss Function: {criterion.__class__.__name__}")
print(f"Optimizer:     {optimizer.__class__.__name__} (lr={learning_rate})")
```

### Module 03 - Cell 3: Implementing the Training Loop
We define the training cycle for our baseline model, performing the forward pass, computing the loss, executing backpropagation, and updating the network weights.

```python
# Number of epochs for initial model training validation
num_epochs = 5

print("--- Starting Training Loop ---")
for epoch in range(num_epochs):
    # Set the model to training mode
    model.train()
    
    # 1. Clear old gradients from the previous step
    optimizer.zero_grad()
    
    # 2. Forward pass: generate predictions 
    # (Using dummy tensors as placeholders for our dataset batch)
    dummy_input = torch.randn(4, 5, 32, 32)
    dummy_target = torch.randn(4, 1, 32, 32)
    
    outputs = model(dummy_input)
    
    # 3. Compute the loss between prediction and target
    loss = criterion(outputs, dummy_target)
    
    # 4. Backward pass: compute gradients via backpropagation
    loss.backward()
    
    # 5. Update model weights using the optimizer
    optimizer.step()
    
    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}")

print("--- Training Loop Verified Successfully ---")
```
